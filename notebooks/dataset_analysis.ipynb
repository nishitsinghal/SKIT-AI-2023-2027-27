```python
from pathlib import Path

# Dataset paths
dementia_path = Path(r"D:\SKIT_Project\dataset\dementia")
nodementia_path = Path(r"D:\SKIT_Project\dataset\nodementia")

# Count audio files
dementia_files = list(dementia_path.rglob("*.wav"))
nodementia_files = list(nodementia_path.rglob("*.wav"))

print("Dementia audio files:", len(dementia_files))
print("No-Dementia audio files:", len(nodementia_files))
print("Total audio files:", len(dementia_files) + len(nodementia_files))
```

```python
print("Dataset Summary")
print("----------------")
print("Dementia samples:", len(dementia_files))
print("No-Dementia samples:", len(nodementia_files))
print("Total samples:", len(dementia_files) + len(nodementia_files))
```

```python
import librosa

# First audio file ko check karte hain
sample_file = dementia_files[0]

audio, sample_rate = librosa.load(sample_file, sr=None)

print("File:", sample_file.name)
print("Sample Rate:", sample_rate, "Hz")
print("Duration:", round(len(audio) / sample_rate, 2), "seconds")
print("Channels: Mono")
```

```python
print("Dataset is ready for further preprocessing.")
print("Total audio files:", len(dementia_files) + len(nodementia_files))
```

```python
import numpy as np

durations = []

for file in dementia_files[:10]:
    audio, sr = librosa.load(file, sr=None)
    durations.append(len(audio) / sr)

print("Sample Audio Analysis")
print("---------------------")
print("Samples checked:", len(durations))
print("Average duration:", round(np.mean(durations), 2), "seconds")
print("Minimum duration:", round(np.min(durations), 2), "seconds")
print("Maximum duration:", round(np.max(durations), 2), "seconds")
```
