```python
import librosa
import soundfile as sf

# Input speech file
input_file = "voice-sample.wav"

# Load audio and convert to 16 kHz mono
audio, sample_rate = librosa.load(
    input_file,
    sr=16000,
    mono=True
)

# Remove silence from beginning and end
processed_audio, _ = librosa.effects.trim(audio)

# Normalize audio
processed_audio = processed_audio / max(abs(processed_audio))

# Save processed audio
sf.write(
    "processed_speech.wav",
    processed_audio,
    sample_rate
)

# Display information
duration = len(processed_audio) / sample_rate

print("Speech Processing Completed Successfully")
print("Sample Rate:", sample_rate, "Hz")
print("Duration:", round(duration, 2), "seconds")
print("Processed file: processed_speech.wav")
```
